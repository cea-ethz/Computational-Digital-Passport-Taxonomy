```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
from copy import deepcopy

# --- Helper Function for Optimization Curve ---
def get_best_so_far(df, score_col):
    """Returns the cumulative maximum series for the score column."""
    df_sorted = df.sort_values('number')
    return df_sorted['number'], df_sorted[score_col].cummax()

# --- Configuration ---
y_min, y_max = -0.1, 1.1
x_max_val = max(
    df_data_points['number'].max(), 
    df_descriptions['number'].max(),
    df_descriptions_gpt['number'].max(),
    df_descriptions_phi['number'].max()
) + 5
x_min, x_max = -3, x_max_val+3

# --- Updated Colorbar Positions ---
cbar_positions = {
    (1, 1): (0.42, 0.8),  # Left Column, Top Row
    (1, 2): (1.03, 0.8),  # Right Column, Top Row
    (2, 1): (0.42, 0.22), # Left Column, Bottom Row
    (2, 2): (1.03, 0.22)  # Right Column, Bottom Row
}

plots_config = [
    (df_data_points, 1, 1, 'Reds', "Data point names"),
    (df_descriptions, 1, 2, 'Blues', "Apertus-8B-Instruct descriptions"),
    (df_descriptions_gpt, 2, 1, 'Greens', "GPT5-nano descriptions"),
    (df_descriptions_phi, 2, 2, 'Oranges', "Phi-4-mini-Instruct descriptions")
]

fig = make_subplots(
    rows=2, cols=2, 
    subplot_titles=[p[4] for p in plots_config],
    horizontal_spacing=0.22, 
    vertical_spacing=0.15 
)

# --- Loop through each dataset ---
i = 0
for df, row, col, cmap, title in plots_config:
    dbcv_scores = deepcopy(df['user_attrs_dbcv_score'])
    dbcv_scores[dbcv_scores < 0] = 0
    
    df['combined_score'] = (df['user_attrs_ccc_score'] + dbcv_scores) / 2
    
    # Get position settings for this subplot
    cbar_x, cbar_y = cbar_positions[(row, col)]

    # 1. Plot the Scatter Points
    fig.add_trace(
        go.Scatter(
            x=df['number'],
            y=df['combined_score'],
            mode='markers',
            marker=dict(
                color=df['user_attrs_n_clusters'],
                colorscale=cmap,
                size=5,
                showscale=True,
                colorbar=dict(
                    title=None, 
                    x=cbar_x, 
                    y=cbar_y,
                    len=0.4, 
                    thickness=15
                )
            ),
            #name=title,
            showlegend=False
        ),
        row=row, col=col
    )

    title_text = f"<b>{title}</b><br>"
    fig.layout.annotations[i].update(text=title_text)

    i += 1
    
    x_bar_annotation = cbar_x - 0.01
    y_bar_annotation = cbar_y + 0.105
    if row == 1 and col == 2:
        x_bar_annotation = cbar_x - 0.01 + 0.015
    elif row == 2:
        y_bar_annotation = 0.13
        if col ==2:
            x_bar_annotation = cbar_x - 0.01 + 0.015

    # 2. Add Custom Rotated Annotation for Colorbar Title
    fig.add_annotation(
        text="Number of clusters",
        x=x_bar_annotation, 
        y=y_bar_annotation,        
        xref="paper", yref="paper",
        textangle=-90,   
        showarrow=False,
        font=dict(size=16, color="black")
    )

    # 3. Plot the Optimization Curve (Overlay)
    x_line, y_line = get_best_so_far(df, 'combined_score')
    
    fig.add_trace(
        go.Scatter(
            x=x_line, y=y_line,
            mode='lines',
            line=dict(color='gray', width=3, dash='dot'), 
            hoverinfo='skip',
            showlegend=False
        ),
        row=row, col=col
    )

    # 4. Find and Annotate the Best Point
    best_row = df.loc[df['combined_score'].idxmax()]
    best_score = best_row['combined_score']
    best_clusters = int(best_row['user_attrs_n_clusters'])
    best_trial = best_row['number']

    # --- Conditional Formatting for the Label ---
    # Default: Label ABOVE the point (ay = -40)
    ay_offset = -40
    
    # If it is the last plot (Phi-4), put Label BELOW the point (ay = 40)
    if row == 2 and col == 2:
        ay_offset = 40

    fig.add_annotation(
        x=best_trial, 
        y=best_score,
        text=f"<b>Best: {best_score:.3f}</b><br>Clusters: {best_clusters}",
        showarrow=True,
        arrowhead=2,
        ax=0, 
        ay=ay_offset, # Use the conditional offset
        bgcolor="rgba(255, 255, 255, 0.8)",
        bordercolor="black",
        borderwidth=1,
        font=dict(size=13, color="black"),
        row=row, col=col
    )

# --- Final Layout ---
fig.update_layout(
    width=900, height=800, 
    plot_bgcolor='white', paper_bgcolor='white',
    font=dict(family='Arial', size=17),
    margin=dict(r=50) 
)

fig.update_xaxes(title_text='Trial Number', range=[x_min, x_max], showgrid=True, gridcolor='lightgray', title_font=dict(size=16))
fig.update_yaxes(title_text='Averaged CCC and DBCV', range=[y_min, y_max], showgrid=True, gridcolor='lightgray', dtick=0.25, title_font=dict(size=16))

fig.write_image("02_260603_optimization_results.png", scale=3)
fig.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np
from copy import deepcopy

# --- Helper Function for Pareto Front ---
def get_pareto_frontier(df_, x_col, y_col):
    """
    Identifies the Pareto frontier for maximization of both x and y.
    Returns the dataframe subset containing only frontier points, sorted by x.
    """
    df = deepcopy(df_)
    # 1. Sort by X descending, then Y descending
    sorted_df = df.sort_values(by=[x_col, y_col], ascending=[False, False])
    
    pareto_front = []
    
    # 2. Iterate and keep points that have a higher Y than any point seen so far
    current_max_y = -np.inf
    for _, row in sorted_df.iterrows():
        if row[y_col] >= current_max_y:
            pareto_front.append(row)
            current_max_y = row[y_col]
            
    return pd.DataFrame(pareto_front).sort_values(by=x_col)

# --- Configuration ---
x_min, x_max = -0.1, 1.1 # CCC Range
y_min, y_max = -0.1, 1.1 # DBCV Range

# --- Updated Colorbar Positions ---
cbar_positions = {
    (1, 1): (0.4, 0.8),  # Left Column, Top Row
    (1, 2): (1.04, 0.8),  # Right Column, Top Row
    (2, 1): (0.4, 0.24), # Left Column, Bottom Row
    (2, 2): (1.04, 0.24)  # Right Column, Bottom Row
}

# --- Custom Color Scales (Start darker than white) ---
# Format: [Start Color, End Color]
# These hex codes ensure the lowest values are visible (approx. 30-40% saturation)
custom_reds =   ['#8dd696',  '#ffcc99', '#a10808', '#224aa1'] # Light Red -> Dark Red a10808
custom_blues =  ['#8dd696',  '#ffcc99', '#a10808', '#224aa1'] # Light Blue -> Dark Blue
custom_greens = ['#8dd696',  '#ffcc99', '#a10808', '#224aa1'] # Light Green -> Dark Green
custom_oranges= ['#8dd696',  '#ffcc99', '#a10808', '#224aa1'] # Light Orange -> Dark Orange

plots_config = [
    (df_data_points, 1, 1, custom_reds, "a) Data point names"),
    (df_descriptions, 1, 2, custom_blues, "b) Apertus-8B-Instruct descriptions"),
    (df_descriptions_gpt, 2, 1, custom_greens, "c) GPT5-nano descriptions"),
    (df_descriptions_phi, 2, 2, custom_oranges, "d) Phi-4-mini-Instruct descriptions")
]

fig = make_subplots(
    rows=2, cols=2, 
    subplot_titles=[p[4] for p in plots_config],
    horizontal_spacing=0.27, 
    vertical_spacing=0.20 
)

# --- Loop through each dataset ---
i = 0
for df_raw, row, col, cmap, title in plots_config:
    df = df_raw.copy()
    
    # Clean DBCV scores
    df['user_attrs_dbcv_score'] = df['user_attrs_dbcv_score'].clip(lower=0)
    df['combined_score'] = (df['user_attrs_ccc_score'] + df['user_attrs_dbcv_score']) / 2
    
    # Get position settings
    cbar_x, cbar_y = cbar_positions[(row, col)]

    max_val = max(df['user_attrs_n_clusters'].max(), 151) # Ensure max is at least >150 so the last bin exists

    default_max_val = 150
    
    # --- 1. Logic Setup ---
    # The exact values where you want specific colors to appear
    if max_val < default_max_val:
        anchors = [0, 25, 75, max_val, max_val]
        default_max_val = max_val
    else:
        anchors = [0, 25, 75, default_max_val, max_val]
    
    # The colors corresponding to those anchors
    # Note: We repeat the last color so the range from 150 to max_val stays Blue.
    gradient_colors = [cmap[0], cmap[1], cmap[2], cmap[3], cmap[3]]
    
    # specific_colors: ['#8dd696', '#ffcc99', '#a10808', '#224aa1', '#224aa1']
    
    # Construct the Continuous Scale
    # We map each anchor value to a normalized position (0 to 1)
    colorscale = [[val / max_val, color] for val, color in zip(anchors, gradient_colors)]

    label_anchors = anchors
    if row == 2 and col == 2:
        label_anchors = [0, 45, 95, default_max_val, max_val]

    fig.add_trace(
        go.Scatter(
            x=df['user_attrs_ccc_score'],
            y=df['user_attrs_dbcv_score'],
            mode='markers',
            # 'range_color' and 'color_continuous_scale' do NOT work here in go.Scatter.
            # They belong inside the marker dictionary.
            marker=dict(
                color=df['user_attrs_n_clusters'],
                colorscale=colorscale,       # Use the custom hard-stop scale
                cmin=0,                      # Replaces range_color[0]
                cmax=max_val,                # Replaces range_color[1]
                size=6,
                opacity=0.9,
                showscale=True, 
                line=dict(width=0.5, color='rgba(100,100,100,0.3)'),
                
                # Consolidate all Colorbar logic here
                colorbar=dict(
                    title=None, 
                    x=cbar_x, 
                    y=cbar_y,
                    len=0.4, 
                    thickness=15,
                    # Define specific ticks to match your bins
                    tickvals=label_anchors,
                    ticktext=["0", "25", "75", f"{default_max_val}", f" "]
                )
            ),
            #hoverinfo='skip', 
            showlegend=False
        ),
        row=row, col=col
    )
    # 2. Calculate Pareto Frontier
    df_pareto = get_pareto_frontier(df, 'user_attrs_ccc_score', 'user_attrs_dbcv_score')

    # 3. Plot Pareto Line (Connecting the dots)
    fig.add_trace(
        go.Scatter(
            x=df_pareto['user_attrs_ccc_score'], 
            y=df_pareto['user_attrs_dbcv_score'],
            mode='lines', 
            line=dict(color='gray', width=2, dash='dot'),
            hoverinfo='skip',
            showlegend=False
        ),
        row=row, col=col
    )


    
    # 4. Plot Pareto Points (HIGHLIGHTED)
    # These are drawn ON TOP of everything else
    fig.add_trace(
        go.Scatter(
            x=df_pareto['user_attrs_ccc_score'],
            y=df_pareto['user_attrs_dbcv_score'],
            mode='markers',
            marker=dict(
                color=df_pareto['user_attrs_n_clusters'],
                colorscale=colorscale,       # Use the custom hard-stop scale
                cmin=0,                      # Replaces range_color[0]
                cmax=max_val,                # Replaces range_color[1]
                size=6, # Larger size
                opacity=1.0, # Full opacity
                line=dict(width=1.5, color='gray'), # Black outline
                showscale=False # No duplicate colorbar
            ),
            customdata=df_pareto[['user_attrs_n_clusters', 'number']],
            hovertemplate="<b>Pareto Point</b><br>CCC: %{x:.3f}<br>DBCV: %{y:.3f}<br>Clusters: %{customdata[0]}<br>Trial: %{customdata[1]}",
            showlegend=False
        ),
        row=row, col=col
    )

    # Update Titles
    title_text = (f"<span style='font-size:19px'>{title}</span><br>"
                  f"<span style='font-size:12px'> </span>")
    fig.layout.annotations[i].update(text=title_text)
    i += 1
    
    # 5. Colorbar Label
    x_bar_annotation = cbar_x - 0.01
    y_bar_annotation = cbar_y + 0.105
    if row == 1 and col == 2:
        x_bar_annotation = cbar_x - 0.01 + 0.015
    elif row == 2:
        y_bar_annotation = 0.13
        if col == 2:
            x_bar_annotation = cbar_x - 0.01 + 0.015

    fig.add_annotation(
        text="Number of clusters",
        x=x_bar_annotation, 
        y=y_bar_annotation,        
        xref="paper", yref="paper",
        textangle=-90,   
        showarrow=False,
        font=dict(size=16, color="black")
    )

    # 6. Annotate "Best Average" Point
    best_row = df_pareto.loc[df_pareto['combined_score'].idxmax()]
    best_ccc = best_row['user_attrs_ccc_score']
    best_dbcv = best_row['user_attrs_dbcv_score']
    best_clusters = int(best_row['user_attrs_n_clusters'])
    
    ay_offset = -75
    ax_offset = -15
    if row == 2 and col == 2:
        ax_offset = -25
        ay_offset = 70

    fig.add_annotation(
        x=best_ccc, 
        y=best_dbcv,
        text=f"CCC: {best_ccc:.2f}<br>DBCV: {best_dbcv:.2f}<br>n_clusters: {best_clusters}",
        showarrow=True,
        arrowhead=2,
        ax=ax_offset, ay=ay_offset, 
        bgcolor="rgba(255, 255, 255, 0.45)",
        #bordercolor="black",
        #borderwidth=1,
        font=dict(size=16, color="black"),
        row=row, col=col
    )

# --- Final Layout ---
fig.update_layout(
    width=900, height=800, 
    plot_bgcolor='white', paper_bgcolor='white',
    font=dict(family='Arial', size=18),
    margin=dict(r=50) 
)

fig.update_xaxes(title_text='CCC Score', range=[x_min, x_max], showgrid=True, gridcolor='lightgray', dtick=0.25, title_font=dict(size=18))
fig.update_yaxes(title_text='DBCV Score', range=[y_min, y_max], showgrid=True, gridcolor='lightgray', dtick=0.25, title_font=dict(size=18))

fig.write_image("260603_pareto_front_optimization.png", scale=3)
fig.show()
```

```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(0)                 # fixed seed -> reproducible CIs
nz  = lambda s: s[s > 0].to_numpy()
base_nz = nz(df_data_points['user_attrs_ccc_score'])

methods = {
    "apertus": df_descriptions['user_attrs_ccc_score'],
    "gpt":     df_descriptions_gpt['user_attrs_ccc_score'],
    "phi":     df_descriptions_phi['user_attrs_ccc_score'],
}

def cliffs_delta(a, b):
    a, b = np.asarray(a), np.asarray(b)
    return np.sign(a[:, None] - b[None, :]).mean()

def hodges_lehmann(a, b):                       # median of all pairwise (a_i - b_j): the shift MWU implies
    return np.median((np.asarray(a)[:, None] - np.asarray(b)[None, :]).ravel())

def welch_mean_ci(a, b, conf=0.95):             # version-proof Welch CI for the mean difference
    a, b = np.asarray(a), np.asarray(b)
    diff = a.mean() - b.mean()
    va, vb, na, nb = a.var(ddof=1), b.var(ddof=1), len(a), len(b)
    se = np.sqrt(va/na + vb/nb)
    df = se**4 / ((va/na)**2/(na-1) + (vb/nb)**2/(nb-1))
    tcrit = stats.t.ppf(1 - (1-conf)/2, df)
    return diff, diff - tcrit*se, diff + tcrit*se

def boot_ci(a, b, stat, conf=0.95, n=5000):     # BCa bootstrap CI for any two-sample statistic
    res = stats.bootstrap((a, b), stat, vectorized=False, paired=False,
                          n_resamples=n, confidence_level=conf,
                          method='BCa', random_state=rng)
    return res.confidence_interval.low, res.confidence_interval.high

print(f"{'method':8s} {'Δmean [95% CI]':>26s} {'δ [95% CI]':>22s} {'p(MWU,1-sided)':>16s}")
for k, v in methods.items():
    v_nz = nz(v)
    dmean, lo_m, hi_m = welch_mean_ci(v_nz, base_nz)
    d   = cliffs_delta(v_nz, base_nz)
    lo_d, hi_d = boot_ci(v_nz, base_nz, cliffs_delta)
    _, p = stats.mannwhitneyu(v_nz, base_nz, alternative='greater')
    print(f"{k:8s} {dmean:+.3f} [{lo_m:+.3f},{hi_m:+.3f}]   "
          f"{d:+.2f} [{lo_d:+.2f},{hi_d:+.2f}]   {p:.2g}")
```
