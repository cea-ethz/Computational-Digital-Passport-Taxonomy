```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np
from copy import deepcopy

# --- Helper Function (Same as before) ---
def get_pareto_frontier(df_, x_col, y_col):
    df = deepcopy(df_)
    sorted_df = df.sort_values(by=[x_col, y_col], ascending=[False, False])
    pareto_front = []
    current_max_y = -np.inf
    for _, row in sorted_df.iterrows():
        if row[y_col] >= current_max_y:
            pareto_front.append(row)
            current_max_y = row[y_col]
    return pd.DataFrame(pareto_front).sort_values(by=x_col)

# --- Configuration ---
x_min, x_max = -0.1, 1.1
y_min, y_max = -0.1, 1.1

# --- Custom Color Scales ---
custom_reds =   ['#8dd696',  '#ffcc99', '#a10808', '#224aa1']
custom_blues =  ['#8dd696',  '#ffcc99', '#a10808', '#224aa1']
custom_greens = ['#8dd696',  '#ffcc99', '#a10808', '#224aa1']
custom_oranges= ['#8dd696',  '#ffcc99', '#a10808', '#224aa1']

plots_config = [
    (df_data_points, 1, 1, custom_reds, "a) Data point names"),
    (df_descriptions, 1, 2, custom_blues, "b) Apertus-8B-Instruct descriptions"),
    (df_descriptions_gpt, 2, 1, custom_greens, "c) GPT5-nano descriptions"),
    (df_descriptions_phi, 2, 2, custom_oranges, "d) Phi-4-mini-Instruct descriptions")
]

# --- 1. NEW 5x5 Grid Strategy ---
# We use Row 3 and Col 3 as "Spacers" to create the large gap.
# The `vertical_spacing` handles the small gap.
fig = make_subplots(
    rows=5, cols=5, 
    # Logic: [Hist, Scatter, SPACER, Hist, Scatter]
    row_heights=[0.10, 0.38, 0.2, 0.10, 0.38], 
    column_widths=[0.38, 0.10, 0.3, 0.38, 0.10],
    
    # TIGHT spacing for Histograms -> Scatter
    horizontal_spacing=0.01, 
    vertical_spacing=0.01, 
    start_cell="top-left"
)

# --- Adjusted Colorbar Positions ---
# Because columns are tighter, we adjust X positions slightly
cbar_positions = {
    (1, 1): (0.42, 0.78),  # Left Col
    (1, 2): (1.05, 0.78),  # Right Col
    (2, 1): (0.42, 0.15), 
    (2, 2): (1.05, 0.15)   
}

# --- Configuration ---
bin_size = 0.05
x_min, x_max = -0.1, 1.1

# Define bin edges exactly as Plotly would use them
bin_edges = np.arange(x_min, x_max + bin_size, bin_size)
bin_config = dict(start=x_min, end=x_max, size=bin_size)


for df_raw, logical_row, logical_col, cmap, title in plots_config:
    df = df_raw.copy()
    
    # --- Map Logical (2x2) to Physical (5x5) Grid ---
    # If logical row is 1 -> Hist Row 1, Scat Row 2
    # If logical row is 2 -> Hist Row 4, Scat Row 5 (Skipping Row 3)
    scat_row = 2 if logical_row == 1 else 5
    hist_x_row = scat_row - 1
    
    # If logical col is 1 -> Scat Col 1, Hist Col 2
    # If logical col is 2 -> Scat Col 4, Hist Col 5 (Skipping Col 3)
    scat_col = 1 if logical_col == 1 else 4
    hist_y_col = scat_col + 1

    # --- Data Prep ---
    df['user_attrs_dbcv_score'] = df['user_attrs_dbcv_score'].clip(lower=0)
    df['combined_score'] = (df['user_attrs_ccc_score'] + df['user_attrs_dbcv_score']) / 2
    
    max_val = max(df['user_attrs_n_clusters'].max(), 151) 
    default_max_val = 150
    anchors = [0, 25, 75, default_max_val, max_val] if max_val >= default_max_val else [0, 25, 75, max_val, max_val]
    gradient_colors = [cmap[0], cmap[1], cmap[2], cmap[3], cmap[3]]
    colorscale = [[val / max_val, color] for val, color in zip(anchors, gradient_colors)]

    cbar_x, cbar_y = cbar_positions[(logical_row, logical_col)]
    label_anchors = anchors
    label_text = ["0", "25", "75", f"{default_max_val}", " "]
    if logical_row == 2 and logical_col == 2:
         label_anchors = [0, 45, 95, default_max_val, max_val] 

    # --- TRACE 1: Top Histogram ---
    fig.add_trace(
        go.Histogram(
            x=df['user_attrs_ccc_score'],
            marker=dict(color='lightgray', line=dict(width=1, color='white')),
            opacity=1,
            xbins=bin_config, autobinx=False,
            showlegend=False
        ), row=hist_x_row, col=scat_col
    )

    fig.update_xaxes(title_text='', showgrid=True, gridcolor='lightgray', dtick=0.25, title_font=dict(size=18), row=hist_x_row, col=scat_col)

    # --- TRACE 2: Right Histogram ---
    fig.add_trace(
        go.Histogram(
            y=df['user_attrs_dbcv_score'],
            marker=dict(color='lightgray', line=dict(width=1, color='white')),
            opacity=1,
            ybins=bin_config, autobiny=False,
            orientation='h',
            showlegend=False,
        ),
        row=scat_row, col=hist_y_col
    )

    # --- TRACE 3: Main Scatter ---
    fig.add_trace(
        go.Scatter(
            x=df['user_attrs_ccc_score'],
            y=df['user_attrs_dbcv_score'],
            mode='markers',
            marker=dict(
                color=df['user_attrs_n_clusters'],
                colorscale=colorscale,
                cmin=0, cmax=max_val,
                size=6, opacity=0.9,
                showscale=True, 
                line=dict(width=0.5, color='rgba(100,100,100,0.3)'),
                colorbar=dict(
                    title=None,
                    x=cbar_x, y=cbar_y,
                    len=0.3, thickness=15,
                    tickvals=label_anchors,
                    ticktext=label_text
                )
            ),
            showlegend=False
        ),
        row=scat_row, col=scat_col
    )

    # --- TRACE 4: Pareto Line ---
    df_pareto = get_pareto_frontier(df, 'user_attrs_ccc_score', 'user_attrs_dbcv_score')
    fig.add_trace(
        go.Scatter(
            x=df_pareto['user_attrs_ccc_score'], 
            y=df_pareto['user_attrs_dbcv_score'],
            mode='lines', 
            line=dict(color='gray', width=2, dash='dot'),
            showlegend=False, hoverinfo='skip'
        ),
        row=scat_row, col=scat_col
    )

    # --- TRACE 5: Pareto Points ---
    fig.add_trace(
        go.Scatter(
            x=df_pareto['user_attrs_ccc_score'],
            y=df_pareto['user_attrs_dbcv_score'],
            mode='markers',
            marker=dict(
                color=df_pareto['user_attrs_n_clusters'],
                colorscale=colorscale, cmin=0, cmax=max_val,
                size=7, opacity=1.0,
                line=dict(width=1.5, color='gray'),
                showscale=False
            ),
            customdata=df_pareto[['user_attrs_n_clusters', 'number']],
            hovertemplate="<b>Pareto</b><br>CCC: %{x:.2f}<br>DBCV: %{y:.2f}<br>Clusters: %{customdata[0]}",
            showlegend=False
        ),
        row=scat_row, col=scat_col
    )

    # --- Titles ---
    fig.add_annotation(
        text=f"{title}",
        row=hist_x_row, col=scat_col,
        x=0, y=1.35, # Pushed up slightly more for clearance
        xref="x domain", yref="y domain",
        xanchor="left", yanchor="bottom",
        showarrow=False,
        font=dict(size=18, color='black')
    )

    # --- "Number of clusters" Label ---
    x_bar_annotation = cbar_x - 0.01
    y_bar_annotation = cbar_y + 0.09
    if logical_row == 1 and logical_col == 2:
        x_bar_annotation = cbar_x - 0.01 + 0.015
    elif logical_row == 2:
        y_bar_annotation = 0.06
        if logical_col == 2:
            x_bar_annotation = cbar_x - 0.01 + 0.015

    fig.add_annotation(
        text="Number of clusters",
        x=x_bar_annotation, 
        y=y_bar_annotation,        
        xref="paper", yref="paper",
        textangle=-90,   
        showarrow=False,
        font=dict(size=16, color="black")
    )


    # --- Best Point Annotation ---
    best_row_val = df_pareto.loc[df_pareto['combined_score'].idxmax()]
    ax_offset = -15 if not (logical_row == 2 and logical_col == 2) else -25
    ay_offset = -50 if not (logical_row == 2 and logical_col == 2) else 50
    fig.add_annotation(
        x=best_row_val['user_attrs_ccc_score'], 
        y=best_row_val['user_attrs_dbcv_score'],
        text=f"CCC: {best_row_val['user_attrs_ccc_score']:.2f}<br>DBCV: {best_row_val['user_attrs_dbcv_score']:.2f}",
        showarrow=True, arrowhead=2,
        ax=ax_offset, ay=ay_offset, 
        bgcolor="rgba(255, 255, 255, 0.6)",
        font=dict(size=16, color="black"),
        row=scat_row, col=scat_col
    )
    
    # --- Axis Titles ---
    if logical_col == 1 or logical_col == 2:
        fig.update_yaxes(title_text='DBCV Score', row=scat_row, col=scat_col)
    if logical_row == 1 or logical_row == 2:
        fig.update_xaxes(title_text='CCC Score', row=scat_row, col=scat_col)

    # --- Force Ranges ---
    fig.update_xaxes(range=[x_min, x_max], row=scat_row, col=scat_col, showgrid=True, gridcolor='lightgray', dtick=0.25)
    fig.update_yaxes(range=[y_min, y_max], row=scat_row, col=scat_col, showgrid=True, gridcolor='lightgray', dtick=0.25)
    
    fig.update_xaxes(range=[x_min, x_max], row=hist_x_row, col=scat_col, showgrid=False, showticklabels=False, gridcolor='lightgray', dtick=0.25)
    fig.update_yaxes(range=[y_min, y_max], row=scat_row, col=hist_y_col, showgrid=False, showticklabels=False, gridcolor='lightgray', dtick=0.25)
    
    fig.update_yaxes(row=hist_x_row, col=scat_col, showticklabels=True, nticks=4) 
    fig.update_xaxes(row=scat_row, col=hist_y_col, showticklabels=True, nticks=4) 

# --- Final Layout ---
fig.update_layout(
    width=1000, height=900, 
    plot_bgcolor='white', paper_bgcolor='white',
    font=dict(family='Arial', size=16),
    margin=dict(l=40, r=40, t=60, b=40),
    showlegend=False
)

fig.write_image("260603_pareto_front_optimization_histograms.png", scale=3)
fig.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np
from scipy import stats
from scipy.stats import ttest_ind
from statsmodels.stats.multitest import multipletests

# --- Configuration ---
bin_size = 0.05
x_min, x_max = -0.1, 1.1
bin_edges = np.arange(x_min, x_max + bin_size, bin_size)
bin_config = dict(start=x_min, end=x_max, size=bin_size)

# --- Data (non-zero = conditional on successful clustering; 0 = degenerate/failed) ---
nz = lambda s: s[s > 0]

raw = {
    "a) Data point names":                 df_data_points['user_attrs_ccc_score'],
    "b) Apertus-8B-Instruct descriptions":  df_descriptions['user_attrs_ccc_score'],
    "c) GPT-5-nano descriptions":           df_descriptions_gpt['user_attrs_ccc_score'],
    "d) Phi-4-mini Instruct descriptions":  df_descriptions_phi['user_attrs_ccc_score'],
}
names = list(raw.keys())
base_full = raw[names[0]]
base_nz   = nz(base_full)

# datasets used for plotting + stats = the NON-ZERO scores (matches the hypothesis)
datasets = [(nz(raw[n]), n, 'lightgrey') for n in names]

def cohens_d(a, b):
    a, b = np.asarray(a), np.asarray(b)
    n1, n2 = len(a), len(b)
    s1, s2 = a.std(ddof=1), b.std(ddof=1)
    pooled = np.sqrt(((n1 - 1) * s1**2 + (n2 - 1) * s2**2) / (n1 + n2 - 2))
    return (a.mean() - b.mean()) / pooled


# --- Hypothesis test: one-sided Welch on non-zero scores, Holm-corrected over the 3 methods ---
method_idx = [1, 2, 3]                       # indices into `names` (everything but the baseline)
raw_p, deltas, fails = {}, {}, {}
for i in method_idx:
    v_nz = nz(raw[names[i]])
    # one-sided Welch: H1 = method > data_points
    _, p_one = ttest_ind(v_nz, base_nz, equal_var=False, alternative='greater')
    raw_p[i]  = p_one
    deltas[i] = cohens_d(v_nz, base_nz)
    fails[i]  = (raw[names[i]] <= 0).mean()

reject, p_holm_arr, *_ = multipletests([raw_p[i] for i in method_idx], method='holm')
p_holm    = dict(zip(method_idx, p_holm_arr))
confirmed = dict(zip(method_idx, reject))
base_fail = (base_full <= 0).mean()

def fmt_p(p):
    return "<0.001" if p < 0.001 else f"{p:.3f}"

# --- Global Y-Max for alignment (computed on the non-zero data we actually plot) ---
# failed (CCC <= 0) counts per group, in the same order as `datasets`
n_fail_counts = [int((raw[n] <= 0).sum()) for n in names]

# Global Y-Max: account for both the non-zero bars AND the failed bars
max_nonzero = max(np.histogram(data, bins=bin_edges)[0].max() for data, _, _ in datasets)
max_freq = max(max_nonzero, max(n_fail_counts))
global_y_max = max_freq + 30


# --- Helper: normal curve (kept for visual reference only; data are non-normal) ---
def get_normal_curve(data, x_min, x_max, bin_size):
    mu, std = data.mean(), data.std()
    x_range = np.linspace(x_min, x_max, 500)
    y_curve = stats.norm.pdf(x_range, mu, std + 1e-9) * len(data) * bin_size
    return x_range, y_curve, mu, std

# --- Subplots ---
fig = make_subplots(rows=2, cols=2, horizontal_spacing=0.15, vertical_spacing=0.22,
                    subplot_titles=["placeholder"] * 4)

for i, (data, name, color) in enumerate(datasets):
    row, col = (i // 2) + 1, (i % 2) + 1
    x_curve, y_curve, mu, std = get_normal_curve(data, x_min, x_max, bin_size)

    fig.layout.annotations[i].update(
        text=f"<span style='font-size:19px'>{name}</span><br><span style='font-size:12px'> </span>")

    fig.add_trace(go.Histogram(x=data, marker_color=color, opacity=1,
                               xbins=bin_config, autobinx=False, showlegend=False),
                  row=row, col=col)
    fig.add_trace(go.Scatter(x=x_curve, y=y_curve, mode='lines',
                             line=dict(color='darkgray', width=2), showlegend=False),
                  row=row, col=col)
    
    # very light grey bar for failed/degenerate trials (CCC = 0), shown at the 0 bin
    n_fail = int((raw[name] <= 0).sum())
    if n_fail > 0:
        fig.add_trace(go.Bar(
            x=[bin_size / 2], y=[n_fail], width=bin_size,
            marker=dict(color='rgba(0,0,0,0.07)', line=dict(width=0)),
            showlegend=False,
            hovertemplate=f'failed (CCC=0): {n_fail}<extra></extra>'
        ), row=row, col=col)

        # fail % label sitting on top of the 0-bar
        fail_pct = n_fail / len(raw[name])
        fig.add_annotation(
            x=bin_size, y=n_fail,
            text=f"Fail: {fail_pct:.0%}",
            showarrow=False, yshift=12,            # 12px above the bar top
            font=dict(color="dark gray", size=15),
            row=row, col=col
        )


        

    fig.add_vline(x=mu, line_width=3, line_dash="dash", line_color="black", row=row, col=col)
    fig.add_vline(x=mu - std, line_width=1, line_dash="dot", line_color="gray", row=row, col=col)
    fig.add_vline(x=mu + std, line_width=1, line_dash="dot", line_color="gray", row=row, col=col)

    if i == 0:   # baseline: no test against itself
        text = f"Mean: {mu:.2f}<br>Std: {std:.2f}<br>(baseline)"
    else:
        tag = "\u2713 confirmed" if raw_p[i] < 0.05 else "n.s."
        text = (f"Mean: {mu:.2f}<br>Std: {std:.2f}"
                f"<br>P-value: {fmt_p(raw_p[i])}<br>Cohen's d: {deltas[i]:+.2f}")

    fig.add_annotation(x=mu, y=max_freq -10, text=text, showarrow=False, yshift=10,
                       font=dict(color="black", size=15),
                       bgcolor="rgba(255, 255, 255, 0.8)", row=row, col=col)

# --- Layout ---
fig.update_layout(width=900, height=900, plot_bgcolor='white', paper_bgcolor='white',
                  font=dict(family='Arial', size=18), bargap=0.05, barmode='overlay')

fig.update_xaxes(title_text='CCC Score', range=[x_min, x_max], showgrid=True,
                 gridcolor='lightgray', dtick=0.25, title_font=dict(size=18))
fig.update_yaxes(title_text='Frequency', range=[0, global_y_max], showgrid=True,
                 gridcolor='lightgray', title_font=dict(size=18))

fig.write_image("260604_optimization_results_ccc.png", scale=3)
fig.show()
```

```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(0)                 # fixed seed -> reproducible CIs
nz  = lambda s: s[s > 0].to_numpy()
base_nz = nz(df_data_points['user_attrs_ccc_score'])

methods = {
    "apertus": df_descriptions['user_attrs_ccc_score'],
    "gpt":     df_descriptions_gpt['user_attrs_ccc_score'],
    "phi":     df_descriptions_phi['user_attrs_ccc_score'],
}

def cliffs_delta(a, b):
    a, b = np.asarray(a), np.asarray(b)
    return np.sign(a[:, None] - b[None, :]).mean()

def hodges_lehmann(a, b):                       # median of all pairwise (a_i - b_j): the shift MWU implies
    return np.median((np.asarray(a)[:, None] - np.asarray(b)[None, :]).ravel())

def welch_mean_ci(a, b, conf=0.95):             # version-proof Welch CI for the mean difference
    a, b = np.asarray(a), np.asarray(b)
    diff = a.mean() - b.mean()
    va, vb, na, nb = a.var(ddof=1), b.var(ddof=1), len(a), len(b)
    se = np.sqrt(va/na + vb/nb)
    df = se**4 / ((va/na)**2/(na-1) + (vb/nb)**2/(nb-1))
    tcrit = stats.t.ppf(1 - (1-conf)/2, df)
    return diff, diff - tcrit*se, diff + tcrit*se

def boot_ci(a, b, stat, conf=0.95, n=5000):     # BCa bootstrap CI for any two-sample statistic
    res = stats.bootstrap((a, b), stat, vectorized=False, paired=False,
                          n_resamples=n, confidence_level=conf,
                          method='BCa', random_state=rng)
    return res.confidence_interval.low, res.confidence_interval.high

print(f"{'method':8s} {'Δmean [95% CI]':>26s} {'δ [95% CI]':>22s} {'p(MWU,1-sided)':>16s}")
for k, v in methods.items():
    v_nz = nz(v)
    dmean, lo_m, hi_m = welch_mean_ci(v_nz, base_nz)
    d   = cliffs_delta(v_nz, base_nz)
    lo_d, hi_d = boot_ci(v_nz, base_nz, cliffs_delta)
    _, p = stats.mannwhitneyu(v_nz, base_nz, alternative='greater')
    print(f"{k:8s} {dmean:+.3f} [{lo_m:+.3f},{hi_m:+.3f}]   "
          f"{d:+.2f} [{lo_d:+.2f},{hi_d:+.2f}]   {p:.2g}")
```
